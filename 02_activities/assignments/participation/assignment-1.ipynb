```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613) # set a random seed for reproducibility
x = np.arange(50) # 0-49 array of integers
y1 = np.random.randint(0,100,50) # 50-element array, range 1-100
y2 = np.random.randint(0,100,50) # duplicate of above
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y1, label = 'Person 1')
ax.plot(x, y2, label = 'Person 2')
ax.legend(loc = 'lower right',
          frameon = True, # this adds a box around the legend
          fontsize = 12, # makes font size 12
          ncol = 2, 
          shadow = True)
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y1, label = 'Person 1')
ax.plot(x, y2, label = 'Person 2')
ax.legend(loc = 'upper left',
          bbox_to_anchor = (1,1)) # this moves the legend to the upper left and away from the plot area
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y1, label = 'Person 1')
ax.plot(x, y2, label = 'Person 2')
ax.legend(loc = 'lower right')
ax.text(10, 95, "This value is important!",
        ha = 'center', # align text
        color = 'red', # make text red
        size = 20
        ) # this places text at the coordinates (10,95) on the plot
fig.show()
```

```python
fig, ax = plt.subplots()
ax.axis([0,10,0,10]) # set the x and y axis limits to 0-10

ax.text(1,5, ". Data:(1,5)", transform=ax.transData) # this places text at the coordinates (1,5) on the plot
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform=ax.transAxes) # this places text at the coordinates (0.5, 0.1) relative to the axes (50% across, 10% up)
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure) # place text 20% across, 20% up in the figure
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y1, label = 'Person 1')
ax.plot(x, y2, label = 'Person 2')
ax.legend(loc = 'lower right')
# the following annotation makes text with an arrow pointing to the coordinates (10,95), text at (20,94)
ax.annotate("This is important!", xy=(10, 95), xytext=(20,94),
            arrowprops=dict(facecolor = 'black'))

fig.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613) # set a random seed for reproducibility
x1 = np.arange(50) # 0-49 array of integers
y1 = np.random.randint(0,75,50) # 50-element array, range 1-75
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"]) # character names
y2 = np.array([110,180,240,99,220]) # duplicate of above
```

```python
bar_colors = ['tab:red', 'tab:blue', 'tab:brown', 'tab:orange', 'tab:green']
fig,(ax1, ax2) = plt.subplots(ncols = 2,
                              nrows = 1,
                              figsize = (7,3))
ax1.scatter(x1, y1, marker='X', color = "forestgreen")
ax2.bar(x2, y2, color = bar_colors)
fig.text(0.5,-0.1,"This is the x label for the whole figure", fontsize = 16, ha = 'center') # this adds a centered x label for the whole figure, not just one subplot
fig.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']],
                                   figsize=(7,4),
                                   layout='constrained') # this lets the xlabels fit
someaxes['ax1'].scatter(x1, y1)
someaxes['ax2'].bar(x2, y2)
someaxes['ax3'].plot(x1, y1)
someaxes['ax1'].set_xlabel('A Big Label', fontsize=16)
someaxes['ax2'].set_xlabel('Another Label', fontsize=16)
someaxes['ax3'].set_xlabel('Label 3: This Time It\'s Personal', fontsize=12)
fig.show()
```

```python
# more sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110,180,240,99,220])
y2 = np.array([170,100,90,120,50])

fig, ax = plt.subplots(figsize=(7,3))

ax.bar(x, y1,
       color = 'indigo')
ax.plot(x, y2,
        color = 'red')
```

```python
y2_sd = np.std(y2) # this calculates the standard deviation of y2.

fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2,
        color = 'red',)
ax.errorbar(x, y2, yerr = y2_sd, fmt = 'none',
            ecolor = 'indigo', elinewidth=4, capsize=6, capthick=4,
            errorevery=2)
# this makes a plot of y2 with error bars (values determined by y2_sd), with purple error bars that have caps.
```

```python
import plotly.graph_objects as go

x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110,180,240,99,220])
```

```python
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
                 on_bad_lines='skip')

df
```

```python
text = " ".join(each for each in df.quote)

wordcloud = WordCloud(background_color='white', colormap='inferno').generate(text)

fig, ax = plt.subplots(figsize=(7,3))
ax.imshow(wordcloud, interpolation='bilinear')
ax.axis('off')
```

```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
```

```python
A = set(['apple', 'banana', 'watermelon'])
B = set(["pumpkin", "blueberry", "apple", "key lime"])

diagram = venn2_unweighted([A, B], 
                           set_labels = ('Fruits', 'Pies'),
                           set_colors=("blue", "red"),
                           alpha=0.5)

diagram.get_label_by_id('10').set_text("\n".join(A-B))
diagram.get_label_by_id('11').set_text("\n".join(A&B))
diagram.get_label_by_id('01').set_text("\n".join(B-A))

plt.show()
```
